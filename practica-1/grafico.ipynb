```python
import pandas as pd
import matplotlib.pyplot as plt

# Leer el CSV separado por punto y coma
df = pd.read_csv("salida.csv", sep=";")

# Limpiar nombres de columnas por si tienen espacios
df.columns = [col.strip() for col in df.columns]

# Crear el gráfico
plt.figure(figsize=(10, 7))

plt.scatter(df["TIMES"], df["double"], color="blue", label="double", marker="o")
plt.scatter(df["TIMES"], df["float"], color="green", label="float", marker="o")

# Etiquetas en cada punto
for _, row in df.iterrows():
    plt.annotate(f'{row["double"]:.4f}',
                 xy=(row["TIMES"], row["double"]),
                 xytext=(-14, 12), textcoords="offset points",
                 color="blue", fontsize=8)
    plt.annotate(f'{row["float"]:.4f}',
                 xy=(row["TIMES"], row["float"]),
                 xytext=(-14, -15), textcoords="offset points",
                 color="green", fontsize=8)

plt.xlabel("TIMES")
plt.ylabel("Tiempo de ejecución (segundos)")
plt.title("Gráfico de TIMES vs double y float")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("grafico.png", dpi=150)
plt.show()
```
